## 1.) Data Ingestion

```python
import os
import langchain_community
from langchain_community.document_loaders import PyPDFLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.vectorstores import FAISS # In-memory and on disk storage vector store
from langchain.prompts import PromptTemplate
from langchain_core.runnables import RunnablePassthrough
```

```python
file_path = os.path.join(os.getcwd(), "data", "sample.pdf")
```

```python
loader = PyPDFLoader(file_path)
documents = loader.load()
```

```python
len(documents)
```

```python
documents[0]
```

```python
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=150, length_function=len)
```

```python
chunked_docs = text_splitter.split_documents(documents=documents)
len(chunked_docs)
```

# Retrieval Process
From the vector database, we are going to fetch or retrieve or rank the most appropriate k result

```python
# In-memory storage of the vector store
# This is useful for testing and small datasets
vector_store = FAISS.from_documents(
    documents=chunked_docs, embedding=embedding_model)
```

```python
relevant_documents = vector_store.similarity_search("llama2 finetuning benchmark experiments",k=10)
relevant_documents
```

```python
retriever = vector_store.as_retriever()
retriever.invoke("llama2 finetuning benchmark experiments", k=10)
```

```python
# Question of the user
#Context: Based on the question, the retriever will fetch the most relevant documents from the vector store.
prompt_template = """
        Answer the question based on the context provided below. 
        If the context does not contain sufficient information, respond with: 
        "I do not have enough information about this."

        Context: {context}

        Question: {question}

        Answer:"""
```

```python
prompt=PromptTemplate(
    template=prompt_template,
    input_variables=["context", "question"]
)
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
parser = StrOutputParser()
```

```python
def format_docs(docs):
    return "\n\n".join([doc.page_content for doc in docs])
```

```python
# rag_chain = prompt | model | parser
rag_chain = (
    {"context":retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | model
    | parser
)
```

```python
rag_chain.invoke("Tell me about the llama2 finetuning benchmark experiments")
```
